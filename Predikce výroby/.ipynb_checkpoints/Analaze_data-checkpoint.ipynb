```python
import itertools
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error
import matplotlib.pyplot as plt
import seaborn as sns
import glob
from pathlib import Path
```

```python
# Daily plot 
DATA_DIR = "forecast_vs_real_1_h_step_30_days_update_data"

files = sorted(glob.glob(f"{DATA_DIR}/*.csv"))

for file in files:
    df = pd.read_csv(file, sep=";", parse_dates=["TimeStamp"])
    df = df.set_index("TimeStamp")
    df = df.asfreq('h') 

    mape = mean_absolute_percentage_error(df["real"], df["forecast"]) * 100
    
    plt.figure(figsize=(10,4))
    plt.plot(df.index, df["real"], label="Real")
    plt.plot(df.index, df["forecast"], label="Forecast")
    plt.title(f"Hybrid Forecast vs Real - {Path(file).stem} | MAPE: {mape:.2f}%")
    plt.xlabel("Time")
    plt.ylabel("Consumption")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Settings
DATA_FOLDER = Path("./forecast_vs_real_1_h_step_24_days_update_data_long")   # change to your folder
FILE_PATTERN = "*.csv"
WINDOWS = [1, 2, 4, 6, 8, 12, 24]


def load_csv(file_path: Path) -> pd.DataFrame:
    df = pd.read_csv(file_path, sep=";")
    df["TimeStamp"] = pd.to_datetime(df["TimeStamp"], errors="coerce")
    df["real"] = pd.to_numeric(df["real"], errors="coerce")
    df["forecast"] = pd.to_numeric(df["forecast"], errors="coerce")

    df = df.dropna(subset=["TimeStamp", "real", "forecast"]).copy()
    df = df.sort_values("TimeStamp").set_index("TimeStamp")
    return df


def mape(y_true: pd.Series, y_pred: pd.Series) -> float:
    mask = y_true != 0
    if mask.sum() == 0:
        return float("nan")
    return ((y_true[mask] - y_pred[mask]).abs() / y_true[mask].abs()).mean() * 100


def compute_window_mape(df: pd.DataFrame, hours: int) -> float:
    if hours == 1:
        return mape(df["real"], df["forecast"])

    agg = df[["real", "forecast"]].resample(f"{hours}h").sum().dropna()
    if agg.empty:
        return float("nan")

    return mape(agg["real"], agg["forecast"])


def process_folder(folder: Path, pattern: str = "*.csv") -> pd.DataFrame:
    files = sorted(folder.glob(pattern))
    if not files:
        raise FileNotFoundError(f"No files found in {folder} matching {pattern}")

    rows = []
    for file_path in files:
        try:
            df = load_csv(file_path)
            row = {"file": file_path.name}
            for w in WINDOWS:
                row[f"{w}h"] = compute_window_mape(df, w)
            rows.append(row)
            print(f"Processed: {file_path.name}")
        except Exception as e:
            print(f"Skipped {file_path.name}: {e}")

    if not rows:
        raise RuntimeError("No valid CSV files were processed.")

    return pd.DataFrame(rows)


def plot_results(results: pd.DataFrame):
    plt.figure(figsize=(10, 6))

    for _, row in results.iterrows():
        y = [row[f"{w}h"] for w in WINDOWS]
        plt.plot(WINDOWS, y, marker="o", alpha=0.7, label=row["file"])

    avg = results[[f"{w}h" for w in WINDOWS]].mean()
    plt.plot(WINDOWS, avg.values, marker="o", linewidth=3, linestyle="--", label="Average")

    plt.xticks(WINDOWS, [f"{w}h" for w in WINDOWS])
    plt.xlabel("Window")
    plt.ylabel("MAPE (%)")
    plt.title("MAPE by aggregation window")
    plt.grid(True, alpha=0.3)
    plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

def plot_daily_mape_from_aggregated_results(results):
    if isinstance(results, str):
        raise TypeError(
            "results is a string. Pass a pandas DataFrame, not a file path.\n"
            "Example:\n"
            "plot_daily_mape_from_aggregated_results(results_df)"
        )

    df = results.copy()

    df["day"] = pd.to_datetime(
        df["file"].astype(str).str.replace(".csv", "", regex=False),
        errors="coerce"
    )
    df = df.sort_values("day")

    window_cols = [c for c in df.columns if str(c).endswith("h")]

    # average MAPE for each window across all days
    avg_mape = df[window_cols].mean().sort_index(
        key=lambda idx: [int(str(x).replace("h", "")) for x in idx]
    )

    fig, axes = plt.subplots(2, 1, figsize=(14, 10), height_ratios=[3, 1.5])

    # Top: daily MAPE lines
    for col in window_cols:
        axes[0].plot(df["day"], df[col], marker="o", label=col)

    axes[0].set_xlabel("Day")
    axes[0].set_ylabel("MAPE (%)")
    axes[0].set_title("Daily MAPE for all windows")
    axes[0].grid(True, alpha=0.3)
    axes[0].legend(title="Window")
    axes[0].tick_params(axis="x", rotation=45)

    # Bottom: average MAPE by window size
    axes[1].plot(avg_mape.index, avg_mape.values, marker="o")
    axes[1].set_xlabel("Window")
    axes[1].set_ylabel("Average MAPE (%)")
    axes[1].set_title("Average MAPE by window size")
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # also return the averages as a Series
    return avg_mape

results = process_folder(DATA_FOLDER, FILE_PATTERN)
plot_daily_mape_from_aggregated_results(results)
```
