```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error

DATA_DIR = Path(".")
OUT_DIR = Path("graphs")
OUT_DIR.mkdir(exist_ok=True)

CSV_SEPARATOR = ";"

FORECAST_PATTERN = "forecast_vs_real_*_h_step_*_days_update_data.csv"
WEATHER_RESULTS_FILE = "wheater_results_allow_no_temp_30d_2.csv"
BASE_DATA_FILE = "energy_with_temp.csv"
```

```python
def load_csv(path: Path) -> pd.DataFrame:
    df = pd.read_csv(path, sep=CSV_SEPARATOR)

    for col in df.columns:
        if "time" in col.lower() or "date" in col.lower() or col.lower() in ["index"]:
            try:
                df[col] = pd.to_datetime(df[col])
            except Exception:
                pass

    return df


def find_column(df: pd.DataFrame, candidates: list[str]) -> str:
    lower_cols = {c.lower(): c for c in df.columns}

    for candidate in candidates:
        if candidate.lower() in lower_cols:
            return lower_cols[candidate.lower()]

    for col in df.columns:
        col_l = col.lower()
        if any(candidate.lower() in col_l for candidate in candidates):
            return col

    raise ValueError(
        f"Nepodařilo se najít sloupec z kandidátů: {candidates}. "
        f"Dostupné sloupce: {list(df.columns)}"
    )


def get_time_column(df: pd.DataFrame) -> str | None:
    candidates = ["TimeStamp", "timestamp", "Datetime", "DateTime", "date", "time", "index"]

    for c in candidates:
        if c in df.columns:
            return c

    for col in df.columns:
        if np.issubdtype(df[col].dtype, np.datetime64):
            return col

    return None


def smape(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    denominator = (np.abs(y_true) + np.abs(y_pred)) / 2
    mask = denominator != 0

    if mask.sum() == 0:
        return np.nan

    return np.mean(np.abs(y_true[mask] - y_pred[mask]) / denominator[mask]) * 100


def metric_table(y_true, y_pred) -> dict:
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    mask = ~np.isnan(y_true) & ~np.isnan(y_pred)
    y_true = y_true[mask]
    y_pred = y_pred[mask]

    errors = y_true - y_pred

    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "MSE": mean_squared_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "MAPE": mean_absolute_percentage_error(y_true, y_pred) * 100,
        "sMAPE": smape(y_true, y_pred),
        "MaxError": np.max(np.abs(errors)),
        "MeanError": np.mean(errors),
        "StdError": np.std(errors),
    }
def save_plot(filename: str):
    plt.tight_layout()
    plt.savefig(OUT_DIR / filename, dpi=300, bbox_inches="tight")
    plt.show()
```

```python
def parse_forecast_filename(path: Path):
    """
    Např.:
    forecast_vs_real_1_h_step_30_days_update_data.csv
    """
    match = re.search(r"forecast_vs_real_(\d+)_h_step_(\d+)_days", path.name)

    if not match:
        return None

    step_hours = int(match.group(1))
    history_days = int(match.group(2))

    train_samples = int((history_days * 24) / step_hours)

    return {
        "file": path.name,
        "step_hours": step_hours,
        "history_days": history_days,
        "train_samples": train_samples,
    }
```

```python
def plot_history_days_by_same_step(results: pd.DataFrame):
    if results.empty:
        print("Nejsou dostupná data.")
        return

    metrics = ["MAE", "RMSE", "sMAPE", "MaxError"]
    available_metrics = [m for m in metrics if m in results.columns]

    for step in sorted(results["step_hours"].unique()):
        step_df = results[results["step_hours"] == step]

        grouped = step_df.groupby("history_days", as_index=False)[available_metrics].mean()
        grouped = grouped.sort_values("history_days")

        for metric in available_metrics:
            plt.figure(figsize=(8, 5))
            plt.plot(grouped["history_days"], grouped[metric], marker="o")
            plt.xlabel("Délka historického okna [dny]")
            plt.ylabel(metric)
            plt.title(f"Vliv délky historického okna na {metric}, časový krok {step} h")
            plt.grid(True)

            save_plot(f"history_days_{metric}_{step}h_step.png")
```

```python
def plot_models_by_train_samples(results: pd.DataFrame, tolerance=0.15):
    """
    Porovnává modely podle počtu trénovacích vzorků.

    tolerance=0.15 znamená, že se porovnají modely,
    které mají počet vzorků v rozmezí ±15 %.
    """

    if results.empty:
        print("Nejsou dostupná data.")
        return

    metrics = ["MAE", "RMSE", "sMAPE", "MaxError"]
    available_metrics = [m for m in metrics if m in results.columns]

    unique_samples = sorted(results["train_samples"].unique())

    for base_samples in unique_samples:
        lower = base_samples * (1 - tolerance)
        upper = base_samples * (1 + tolerance)

        subset = results[
            (results["train_samples"] >= lower) &
            (results["train_samples"] <= upper)
        ].copy()

        if subset["step_hours"].nunique() < 2:
            continue

        subset["label"] = (
            subset["step_hours"].astype(str) + "h, " +
            subset["history_days"].astype(str) + " dní"
        )

        for metric in available_metrics:
            plot_df = subset.sort_values(metric)

            plt.figure(figsize=(10, 5))
            plt.bar(plot_df["label"], plot_df[metric])
            plt.xlabel("Varianta modelu")
            plt.ylabel(metric)
            plt.title(
                f"Porovnání modelů s podobným počtem vzorků "
                f"({int(lower)}–{int(upper)} vzorků)"
            )
            plt.xticks(rotation=30, ha="right")
            plt.grid(axis="y")

            save_plot(f"similar_train_samples_{base_samples}_{metric}.png")
```

```python
def save_metrics_table(results: pd.DataFrame):
    if results.empty:
        print("Tabulka metrik je prázdná.")
        return

    metric_cols = [
        "MAE",
        "MSE",
        "RMSE",
        "MAPE",
        "sMAPE",
        "MaxError",
        "MeanError",
        "StdError"
    ]

    available_cols = [col for col in metric_cols if col in results.columns]

    table = results[
        ["file", "step_hours", "history_days"] + available_cols
    ].copy()

    table = table.sort_values(["step_hours", "history_days"])

    table.to_csv(
        OUT_DIR / "all_metrics_table.csv",
        sep=CSV_SEPARATOR,
        index=False
    )

    display(table)

    return table
```

```python
def plot_history_window_all_metrics(results: pd.DataFrame):
    if results.empty:
        print("Nejsou dostupná data pro porovnání metrik.")
        return

    metrics = ["MAE", "RMSE", "sMAPE", "MaxError"]
    available_metrics = [m for m in metrics if m in results.columns]

    grouped = results.groupby("history_days", as_index=False)[available_metrics].mean()

    for metric in available_metrics:
        plt.figure(figsize=(8, 5))
        plt.plot(grouped["history_days"], grouped[metric], marker="o")
        plt.xlabel("Délka historického okna [dny]")
        plt.ylabel(metric)
        plt.title(f"Vliv délky historického okna na metriku {metric}")
        plt.grid(True)
        save_plot(f"history_window_{metric}.png")
```

```python
def plot_step_all_metrics(results: pd.DataFrame):
    if results.empty:
        print("Nejsou dostupná data pro porovnání metrik.")
        return

    metrics = ["MAE", "RMSE", "sMAPE", "MaxError"]
    available_metrics = [m for m in metrics if m in results.columns]

    grouped = results.groupby("step_hours", as_index=False)[available_metrics].mean()

    for metric in available_metrics:
        plt.figure(figsize=(8, 5))
        plt.bar(grouped["step_hours"].astype(str), grouped[metric])
        plt.xlabel("Časový krok predikce [h]")
        plt.ylabel(metric)
        plt.title(f"Porovnání časového kroku podle metriky {metric}")
        plt.grid(axis="y")
        save_plot(f"step_hours_{metric}.png")
```

```python
def plot_error_by_hour_all_metrics(csv_file: Path):
    df = load_csv(csv_file)

    time_col = get_time_column(df)

    if time_col is None:
        print("Nebyl nalezen časový sloupec.")
        return

    real_col = find_column(df, ["real", "actual", "skutecna", "skutečná", "P_HOME", "y_true"])
    pred_col = find_column(df, ["forecast", "prediction", "predicted", "predikce", "y_pred"])

    df["hour"] = pd.to_datetime(df[time_col]).dt.hour
    df["error"] = df[real_col] - df[pred_col]
    df["abs_error"] = np.abs(df["error"])
    df["squared_error"] = df["error"] ** 2

    grouped = df.groupby("hour").agg(
        MAE=("abs_error", "mean"),
        RMSE=("squared_error", lambda x: np.sqrt(np.mean(x))),
        MaxError=("abs_error", "max"),
        MeanError=("error", "mean")
    ).reset_index()

    grouped.to_csv(
        OUT_DIR / "error_by_hour_metrics.csv",
        sep=CSV_SEPARATOR,
        index=False
    )

    for metric in ["MAE", "RMSE", "MaxError", "MeanError"]:
        plt.figure(figsize=(9, 5))
        plt.bar(grouped["hour"], grouped[metric])
        plt.xlabel("Hodina dne")
        plt.ylabel(metric)
        plt.title(f"Chyba predikce podle hodiny dne ({metric})")
        plt.grid(axis="y")
        save_plot(f"error_by_hour_{metric}.png")

    display(grouped)

    return grouped
```

```python
def plot_forecast_vs_real(csv_file: Path, output_name="forecast_vs_real.png"):
    df = load_csv(csv_file)

    time_col = get_time_column(df)
    real_col = find_column(df, ["real", "actual", "skutecna", "skutečná", "P_HOME", "y_true"])
    pred_col = find_column(df, ["forecast", "prediction", "predicted", "predikce", "y_pred"])

    x = df[time_col] if time_col else df.index

    plt.figure(figsize=(12, 5))
    plt.plot(x, df[real_col], label="Skutečná spotřeba")
    plt.plot(x, df[pred_col], label="Predikovaná spotřeba")
    plt.xlabel("Čas")
    plt.ylabel("Spotřeba")
    plt.title("Porovnání skutečné a predikované spotřeby")
    plt.legend()
    plt.grid(True)

    save_plot(output_name)
```

```python
def plot_residuals(csv_file: Path, output_name="forecast_residuals.png"):
    df = load_csv(csv_file)

    time_col = get_time_column(df)
    real_col = find_column(df, ["real", "actual", "skutecna", "skutečná", "P_HOME", "y_true"])
    pred_col = find_column(df, ["forecast", "prediction", "predicted", "predikce", "y_pred"])

    residuals = df[real_col] - df[pred_col]
    x = df[time_col] if time_col else df.index

    plt.figure(figsize=(12, 5))
    plt.plot(x, residuals, label="Rezidua")
    plt.axhline(0, linestyle="--", linewidth=1)
    plt.xlabel("Čas")
    plt.ylabel("Chyba predikce")
    plt.title("Rezidua predikčního modelu")
    plt.legend()
    plt.grid(True)

    save_plot(output_name)


def plot_residual_histogram(csv_file: Path, output_name="forecast_residuals_histogram.png"):
    df = load_csv(csv_file)

    real_col = find_column(df, ["real", "actual", "skutecna", "skutečná", "P_HOME", "y_true"])
    pred_col = find_column(df, ["forecast", "prediction", "predicted", "predikce", "y_pred"])

    residuals = df[real_col] - df[pred_col]

    plt.figure(figsize=(8, 5))
    plt.hist(residuals.dropna(), bins=30)
    plt.xlabel("Chyba predikce")
    plt.ylabel("Četnost")
    plt.title("Rozdělení reziduí modelu")
    plt.grid(True)

    save_plot(output_name)
```

```python
def parse_forecast_filename(path: Path):
    match = re.search(r"forecast_vs_real_(\d+)_h_step_(\d+)_days", path.name)

    if not match:
        return None

    return {
        "file": path.name,
        "step_hours": int(match.group(1)),
        "history_days": int(match.group(2)),
    }


def evaluate_forecast_files():
    rows = []

    for file in DATA_DIR.glob(FORECAST_PATTERN):
        parsed = parse_forecast_filename(file)

        if parsed is None:
            continue

        df = load_csv(file)

        try:
            real_col = find_column(df, ["real", "actual", "skutecna", "skutečná", "P_HOME", "y_true"])
            pred_col = find_column(df, ["forecast", "prediction", "predicted", "predikce", "y_pred"])
        except ValueError:
            continue

        metrics = metric_table(df[real_col], df[pred_col])

        rows.append({
            **parsed,
            **metrics
        })

    results = pd.DataFrame(rows)

    if not results.empty:
        results.to_csv(OUT_DIR / "model_window_step_metrics.csv", sep=CSV_SEPARATOR, index=False)

    return results
```

```python
def plot_step_comparison(results: pd.DataFrame):
    if results.empty:
        print("Nejsou dostupná data pro porovnání časového kroku.")
        return

    grouped = results.groupby("step_hours", as_index=False)["RMSE"].mean()

    plt.figure(figsize=(8, 5))
    plt.bar(grouped["step_hours"].astype(str), grouped["RMSE"])
    plt.xlabel("Časový krok predikce [h]")
    plt.ylabel("RMSE")
    plt.title("Porovnání přesnosti pro různé časové kroky")
    plt.grid(axis="y")

    save_plot("step_hours_comparison_rmse.png")


def plot_history_window_comparison(results: pd.DataFrame):
    if results.empty:
        print("Nejsou dostupná data pro porovnání délky historického okna.")
        return

    grouped = results.groupby("history_days", as_index=False)["RMSE"].mean()

    plt.figure(figsize=(8, 5))
    plt.plot(grouped["history_days"], grouped["RMSE"], marker="o")
    plt.xlabel("Délka historického okna [dny]")
    plt.ylabel("RMSE")
    plt.title("Vliv délky historického okna na přesnost predikce")
    plt.grid(True)

    save_plot("history_window_comparison_rmse.png")
```

```python
def plot_weather_feature_comparison():
    path = DATA_DIR / WEATHER_RESULTS_FILE

    if not path.exists():
        print(f"Soubor {WEATHER_RESULTS_FILE} nebyl nalezen.")
        return

    df = load_csv(path)

    if "combo" not in df.columns:
        print(f"Soubor neobsahuje sloupec combo. Sloupce: {list(df.columns)}")
        return

    if "mean_mse" in df.columns:
        df["RMSE"] = np.sqrt(df["mean_mse"])
        metric_col = "RMSE"
    elif "RMSE" in df.columns:
        metric_col = "RMSE"
    elif "mean_mae" in df.columns:
        metric_col = "mean_mae"
    elif "MAE" in df.columns:
        metric_col = "MAE"
    else:
        print(f"Nebyl nalezen sloupec s metrikou. Sloupce: {list(df.columns)}")
        return

    df_plot = (
        df[["combo", metric_col]]
        .dropna()
        .groupby("combo", as_index=False)[metric_col]
        .mean()
        .sort_values(metric_col)
    )

    display(df_plot)

    plt.figure(figsize=(10, 5))
    plt.bar(df_plot["combo"].astype(str), df_plot[metric_col])
    plt.xlabel("Použité meteorologické veličiny")
    plt.ylabel(metric_col)
    plt.title("Vliv meteorologických veličin na přesnost predikce")
    plt.xticks(rotation=35, ha="right")
    plt.grid(axis="y")

    save_plot("weather_features_comparison_fixed.png")
```

```python
def plot_xgboost_feature_importance():
    path = DATA_DIR / BASE_DATA_FILE

    if not path.exists():
        print(f"Soubor {BASE_DATA_FILE} nebyl nalezen.")
        return

    try:
        import xgboost as xgb
    except ImportError:
        print("Balíček xgboost není nainstalován.")
        return

    df = load_csv(path)

    time_col = get_time_column(df)

    if time_col is not None:
        df = df.set_index(time_col).sort_index()

    value_col = find_column(df.reset_index(), ["P_HOME", "consumption", "spotreba", "spotřeba", "value"])
    temp_col = find_column(df.reset_index(), ["Temperature", "temperature", "temp"])

    df[value_col] = pd.to_numeric(df[value_col], errors="coerce")
    df[temp_col] = pd.to_numeric(df[temp_col], errors="coerce")

    df["hour"] = df.index.hour
    df["is_weekend"] = (df.index.dayofweek >= 5).astype(int)
    df["lag_24"] = df[value_col].shift(24)
    df["lag_48"] = df[value_col].shift(48)
    df["lag_168"] = df[value_col].shift(168)

    feature_cols = ["hour", "is_weekend", "lag_24", "lag_48", "lag_168", temp_col]

    model_df = df[[value_col] + feature_cols].dropna()

    X = model_df[feature_cols]
    y = model_df[value_col]

    model = xgb.XGBRegressor(
        objective="reg:squarederror",
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        random_state=42
    )

    model.fit(X, y)

    importance = pd.DataFrame({
        "feature": feature_cols,
        "importance": model.feature_importances_
    }).sort_values("importance", ascending=True)

    importance.to_csv(OUT_DIR / "xgboost_feature_importance.csv", sep=CSV_SEPARATOR, index=False)

    plt.figure(figsize=(8, 5))
    plt.barh(importance["feature"], importance["importance"])
    plt.xlabel("Důležitost proměnné")
    plt.ylabel("Vstupní proměnná")
    plt.title("Důležitost vstupních proměnných modelu XGBoost")
    plt.grid(axis="x")

    save_plot("xgboost_feature_importance.png")
```

```python
def plot_error_by_hour(csv_file: Path, output_name="error_by_hour.png"):
    df = load_csv(csv_file)

    time_col = get_time_column(df)

    if time_col is None:
        print("Nebyl nalezen časový sloupec.")
        return

    real_col = find_column(df, ["real", "actual", "skutecna", "skutečná", "P_HOME", "y_true"])
    pred_col = find_column(df, ["forecast", "prediction", "predicted", "predikce", "y_pred"])

    df["hour"] = pd.to_datetime(df[time_col]).dt.hour
    df["abs_error"] = (df[real_col] - df[pred_col]).abs()

    grouped = df.groupby("hour", as_index=False)["abs_error"].mean()

    plt.figure(figsize=(9, 5))
    plt.bar(grouped["hour"], grouped["abs_error"])
    plt.xlabel("Hodina dne")
    plt.ylabel("Průměrná absolutní chyba")
    plt.title("Průměrná chyba predikce podle hodiny dne")
    plt.grid(axis="y")

    save_plot(output_name)
```

```python
forecast_files = sorted(DATA_DIR.glob(FORECAST_PATTERN))

if not forecast_files:
    print("Nebyly nalezeny forecast_vs_real CSV soubory.")
else:
    selected_file = forecast_files[0]
    print(f"Použit hlavní soubor: {selected_file.name}")

    plot_forecast_vs_real(selected_file)
    plot_residuals(selected_file)
    plot_residual_histogram(selected_file)
    plot_error_by_hour_all_metrics(selected_file)

results = evaluate_forecast_files()

display(results.sort_values(["step_hours", "history_days"]))

metrics_table = save_metrics_table(results)

plot_weather_feature_comparison()

plot_history_days_by_same_step(results)

plot_models_by_train_samples(results, tolerance=0.15)

print(f"Grafy a tabulky byly uloženy do složky: {OUT_DIR.resolve()}")
```

```python
def clean_label(label):
    if label == "[]":
        return "bez počasí"

    mapping = {
        "temperature": "teplota",
        "wind_speed": "vítr",
        "pressure": "tlak",
        "precipitation": "srážky"
    }

    label = label.replace("[", "").replace("]", "").replace("'", "")
    parts = [p.strip() for p in label.split(",")]
    parts = [mapping.get(p, p) for p in parts]

    return " + ".join(parts)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = "wheater_all_forecasts_allow_no_temp_30d_2.csv"

df = pd.read_csv(file_path, sep=";")

df["error"] = df["actual"] - df["forecast"]
df["abs_error"] = np.abs(df["error"])
df["squared_error"] = df["error"] ** 2

summary = (
    df.groupby("weather_combo")
    .agg(
        RMSE=("squared_error", lambda x: np.sqrt(np.mean(x))),
        MAE=("abs_error", "mean"),
        MeanError=("error", "mean")
    )
    .reset_index()
)

summary["label"] = summary["weather_combo"].apply(clean_label)
summary = summary.sort_values("RMSE")

baseline = summary.loc[summary["weather_combo"] == "[]", "RMSE"].iloc[0]
temperature_only = summary.loc[summary["weather_combo"] == "['temperature']", "RMSE"].iloc[0]

plt.figure(figsize=(13, 6))

bars = plt.bar(summary["label"], summary["RMSE"])

for i, combo in enumerate(summary["weather_combo"]):
    if combo == "[]":
        bars[i].set_edgecolor("black")
        bars[i].set_linewidth(2.5)
    elif "temperature" in combo:
        bars[i].set_hatch("//")

plt.axhline(baseline, linestyle="--", linewidth=2, label="referenční model (bez počasí)")
plt.axhline(temperature_only, linestyle=":", linewidth=2, label="model s teplotou")

plt.title("Vliv meteorologických veličin na přesnost predikce")
plt.xlabel("Použité meteorologické veličiny")
plt.ylabel("RMSE")
plt.xticks(rotation=35, ha="right")
plt.legend()
plt.ylim(950, 1030)
plt.tight_layout()
plt.savefig("weather_rmse_comparison_highlighted.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
summary["delta_RMSE"] = summary["RMSE"] - baseline

plt.figure(figsize=(13, 6))

bars = plt.bar(summary["label"], summary["delta_RMSE"])

for i, combo in enumerate(summary["weather_combo"]):
    if combo == "[]":
        bars[i].set_edgecolor("black")
        bars[i].set_linewidth(2.5)
    elif "temperature" in combo:
        bars[i].set_hatch("//")

plt.axhline(0, linestyle="--", linewidth=2)

plt.title("Změna RMSE oproti modelu bez meteorologických vstupů")
plt.xlabel("Použité meteorologické veličiny")
plt.ylabel("Rozdíl RMSE")

plt.xticks(rotation=35, ha="right")

plt.tight_layout()
plt.savefig("weather_delta_rmse_vs_baseline.png", dpi=300, bbox_inches="tight")
plt.show()
```
