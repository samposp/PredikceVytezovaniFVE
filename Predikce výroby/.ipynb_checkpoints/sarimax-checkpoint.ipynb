```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
```

```python
# -----------------------------
# Configuration
# -----------------------------
DB_PATH = "FVEDB.db"
CSV_PATH = "Ibehej.csv"
TABLE_NAME = "MqttData"
DATE_COL = "Date"
TIME_COL = "Time"
VALUE_COL = "P_HOME"

TIMESTAMP_COL = "TimeStamp"

FORECAST_HOURS = 24 

# SARIMA parameters
ORDER = (1, 1, 1)
SEASONAL_ORDER = (1, 1, 1, 24)  # 24-hour daily seasonality
```

```python
# -----------------------------
# Load data from SQLite
# -----------------------------
conn = sqlite3.connect(DB_PATH)

query = f"""
SELECT {DATE_COL}, {TIME_COL}, {VALUE_COL}
FROM {TABLE_NAME}
ORDER BY {DATE_COL}, {TIME_COL}
"""

df = pd.read_sql_query(query, conn)
conn.close()

df[TIMESTAMP_COL] = pd.to_datetime(df[DATE_COL] + " " + df[TIME_COL], format="%d.%m.%Y %H:%M:%S")
df = df.set_index(TIMESTAMP_COL)
df = df.drop(DATE_COL, axis='columns')
df = df.drop(TIME_COL, axis='columns')
```

```python
# Load CSV
dfscv = pd.read_csv("energy_with_temp.csv", sep=";")
# Convert timestamp
df = pd.DataFrame()
df[TIMESTAMP_COL] = pd.to_datetime(dfscv[TIMESTAMP_COL])
df[VALUE_COL] = dfscv[VALUE_COL]
df["temperature"] = dfscv["temperature"]
# Set time index
df = df.set_index(TIMESTAMP_COL)
```

```python
# -----------------------------
# Prepare time series
# -----------------------------

# ensure hourly frequency
df = df.resample("h").mean()

# fill missing values
df[VALUE_COL] = df[VALUE_COL].interpolate()

series = df[VALUE_COL]
```

```python
# -----------------------------
# Plot original data
# -----------------------------
plt.figure(figsize=(12,6))

trimdf = df.loc["2025-10-19":"2025-12-19"].copy()
# trimdf[VALUE_COL] = pd.to_numeric(trimdf[VALUE_COL], errors="coerce")
# trimdf = trimdf.dropna(subset=[VALUE_COL])

# trimdf = df
series = trimdf[VALUE_COL]

trimdf["day_of_week"] = trimdf.index.dayofweek  # 0 = Monday, 6 = Sunday
# Optional: one-hot encode day of week
exog = pd.get_dummies(trimdf[["day_of_week", "temperature"]], columns=["day_of_week"], drop_first=True)
# exog = exog.apply(pd.to_numeric, errors="coerce")
# exog = exog.fillna(method="ffill")
exog = exog.astype(float)
exog = trimdf["temperature"]

print(series.dtypes)
print(exog.dtypes)
print(series.shape, exog.shape)

plt.plot(trimdf.index, trimdf[VALUE_COL])
plt.title("Original Household Energy Consumption")
plt.xlabel("Time")
plt.ylabel("Consumption")
plt.grid(True)

plt.figure(figsize=(12,6))
plt.plot(trimdf.index, trimdf["temperature"])


plt.show()

```

```python
from pmdarima import auto_arima
auto_model = auto_arima(
    series,
    exogenous=exog,
    seasonal=True,
    m=24,               # daily seasonality
    stepwise=True,
    trace=True,
    error_action='ignore',
    suppress_warnings=True
)

print("Best SARIMAX order found:", auto_model.order, auto_model.seasonal_order)
```

```python
# -----------------------------
# Train SARIMA model
# -----------------------------
model = SARIMAX(
    series,
    # exog=exog,
    # order=ORDER,
    order=(2, 0, 3),
    # seasonal_order=SEASONAL_ORDER,
    seasonal_order=(1, 0, 1, 24),
    enforce_stationarity=False,
    enforce_invertibility=False
)

results = model.fit()

print(results.summary())
```

```python
# -----------------------------
# Forecast
# -----------------------------

# Prepare future exogenous variables
future_index = pd.date_range(
    start=series.index[-1] + pd.Timedelta(hours=1),
    periods=FORECAST_HOURS,
    freq="h"
)

# Create future exogenous DataFrame
future_exog = pd.DataFrame({
    "temperature": df["temperature"].reindex(future_index).fillna(method="ffill"),
    # "day_of_week": future_index.dayofweek
}, index=future_index)
# future_exog = pd.get_dummies(future_exog, columns=["day_of_week"], drop_first=True)
# future_exog = future_exog.astype(float)

# for col in exog.columns:
#     if col not in future_exog.columns:
#         future_exog[col] = 0
# future_exog = future_exog[exog.columns]


# forecast = results.get_forecast(steps=FORECAST_HOURS, exog=future_exog)
forecast = results.get_forecast(steps=FORECAST_HOURS)

forecast_values = forecast.predicted_mean
forecast_ci = forecast.conf_int()

# -----------------------------
# Build forecast dataframe
# -----------------------------
# forecast_index = pd.date_range(
#     start=series.index[-1] + pd.Timedelta(hours=1),
#     periods=FORECAST_HOURS,
#     freq="h"
# )

forecast_df = pd.DataFrame({
    "forecast": forecast_values.values,
}, index=future_index)
```

```python

day = df.index.min().strftime("%Y-%m-%d")
# real data
real_day = df.loc[day]

# forecast data
forecast_day = forecast_df.loc[day]

# merge them
comparison = pd.concat([real_day[VALUE_COL], forecast_day["forecast"]], axis=1)
comparison.columns = ["real", "forecast"]

# percentage error per hour
comparison["perc_error"] = (
    (comparison["real"] - comparison["forecast"]).abs()
    / comparison["real"]
) * 100

mape = comparison["perc_error"].mean()
print(f"Average percentage error: {mape:.2f}%")

plt.figure(figsize=(12,6))

plt.plot(comparison.index, comparison["real"], label="Real")
plt.plot(comparison.index, comparison["forecast"], label="Forecast")

plt.title(f"Real vs Forecast ({day})")
plt.xlabel("Time")
plt.ylabel("Consumption")

plt.legend()
plt.grid(True)

plt.figure(figsize=(12,6))
plt.plot(real_day.index, real_day["temperature"])

# Sum of real energy for the day
real_total = df.loc[day][VALUE_COL].sum()

# Sum of forecasted energy for the day
forecast_total = forecast_df.loc[day]["forecast"].sum()

# Calculate absolute error
abs_error = abs(real_total - forecast_total)

# Calculate percentage error
perc_error = (abs_error / real_total) * 100

# Print results
print(f"Real total energy: {real_total:.2f}")
print(f"Forecast total energy: {forecast_total:.2f}")
print(f"Absolute error: {abs_error:.2f}")
print(f"Percentage error: {perc_error:.2f}%")
```

```python
import requests
start = df.index.min().strftime("%Y-%m-%d")
end = df.index.max().strftime("%Y-%m-%d")

url = (
    "https://archive-api.open-meteo.com/v1/archive"
    f"?latitude=50.793"
    f"&longitude=15.138"
    f"&start_date={start}"
    f"&end_date={end}"
    f"&hourly=temperature_2m"
    f"&timezone=auto"
)

response = requests.get(url)
data = response.json()

# Convert API data to dataframe
temp_df = pd.DataFrame({
    "timestamp": data["hourly"]["time"],
    "temperature": data["hourly"]["temperature_2m"]
})
temp_df["timestamp"] = pd.to_datetime(temp_df["timestamp"])
temp_df = temp_df.set_index("timestamp")
df = df.join(temp_df)
df.to_csv("energy_with_temp.csv", sep=";")
```
