# OTE PublicDataService → SQLite (DateTime, Value only)

This notebook downloads public OTE day-ahead market electricity data from the public SOAP service and stores only:

- `dt`
- `value`

into SQLite.


```python

import requests
import sqlite3
import pandas as pd
from datetime import datetime, timedelta
from xml.etree import ElementTree as ET
from pathlib import Path
from typing import Optional, List, Dict

SERVICE_URL = "http://www.ote-cr.cz/services/PublicDataService"
SOAP_NS = "http://schemas.xmlsoap.org/soap/envelope/"
PUB_NS = "http://www.ote-cr.cz/schema/service/public"

START_DATE = "2025-11-01"   # inclusive
END_DATE   = "2026-04-07"   # inclusive
IN_EUR = True

SQLITE_PATH = Path("FVEDB.db")
TABLE_NAME = "SpotData"

REQUEST_TIMEOUT = 60

print("Service URL:", SERVICE_URL)
print("Date range :", START_DATE, "->", END_DATE)
print("SQLite DB  :", SQLITE_PATH.resolve())
```

```python

def build_get_dam_price_period_envelope(start_date: str, end_date: str, in_eur: bool = True) -> str:
    in_eur_str = "true" if in_eur else "false"
    return f'''<?xml version="1.0" encoding="utf-8"?>
<soapenv:Envelope xmlns:soapenv="{SOAP_NS}" xmlns:pub="{PUB_NS}">
  <soapenv:Header/>
  <soapenv:Body>
    <pub:GetDamPricePeriodE>
      <pub:StartDate>{start_date}</pub:StartDate>
      <pub:EndDate>{end_date}</pub:EndDate>
      <pub:PeriodResolution>PT15M</pub:PeriodResolution>
    </pub:GetDamPricePeriodE>
  </soapenv:Body>
</soapenv:Envelope>'''

def call_get_dam_price_period(start_date: str, end_date: str, in_eur: bool = True) -> str:
    xml_body = build_get_dam_price_period_envelope(start_date, end_date, in_eur)
    headers = {
        "Content-Type": "text/xml; charset=utf-8",
        "SOAPAction": '""',
    }
    resp = requests.post(SERVICE_URL, data=xml_body.encode("utf-8"), headers=headers, timeout=REQUEST_TIMEOUT)
    resp.raise_for_status()
    return resp.text
```

```python
def strip_ns(tag: str) -> str:
    return tag.split("}", 1)[1] if "}" in tag else tag

def parse_float(value):
    if value is None:
        return None
    value = value.strip().replace(",", ".")
    try:
        return float(value)
    except Exception:
        return None

def parse_to_simple_df(xml_text: str) -> pd.DataFrame:
    root = ET.fromstring(xml_text.encode("utf-8"))
    rows = []

    for elem in root.iter():
        if strip_ns(elem.tag) != "Item":
            continue

        children = {
            strip_ns(ch.tag): (ch.text.strip() if ch.text else None)
            for ch in list(elem)
        }

        delivery_date = children.get("Date")
        period_interval = children.get("PeriodInterval")
        value = parse_float(children.get("Price"))

        # skip rows without usable price
        if not delivery_date or not period_interval or value is None:
            continue

        # example: "00:15-00:30" -> take start time
        start_time = period_interval.split("-")[0]

        dt = datetime.fromisoformat(f"{delivery_date}T{start_time}:00").isoformat(sep=" ")

        rows.append({
            "DateTime": dt,
            "Value": value,
        })

    df = pd.DataFrame(rows)
    if not df.empty:
        df = df.drop_duplicates(subset=["DateTime"]).sort_values("DateTime").reset_index(drop=True)

    return df
```

```python


def upsert_dataframe(df: pd.DataFrame, db_path: Path, table_name: str = TABLE_NAME):
    if df.empty:
        return 0
    sql = f"INSERT OR REPLACE INTO {table_name} (DateTime, Value) VALUES (?, ?)"
    with sqlite3.connect(db_path) as conn:
        conn.executemany(sql, df[["DateTime", "Value"]].itertuples(index=False, name=None))
        conn.commit()
    return len(df)

def delete_period_from_db(start_datetime: str, end_datetime: str, db_path: Path, table_name: str = TABLE_NAME):
    sql = f'''
        DELETE FROM {table_name}
        WHERE DateTime >= ? AND DateTime <= ?
    '''

    with sqlite3.connect(db_path) as conn:
        cur = conn.execute(sql, (start_datetime, end_datetime))
        conn.commit()
    return cur.rowcount
```

```python
from datetime import date, timedelta
def split_date_range(start_date: str, end_date: str, max_days: int = 31):
    start = date.fromisoformat(start_date)
    end = date.fromisoformat(end_date)

    ranges = []
    current = start

    while current <= end:
        chunk_end = min(current + timedelta(days=max_days - 1), end)
        ranges.append((current.isoformat(), chunk_end.isoformat()))
        current = chunk_end + timedelta(days=1)

    return ranges

def load_dam_price_period_range(start_date: str, end_date: str, in_eur: bool = True) -> pd.DataFrame:
    all_parts = []

    for chunk_start, chunk_end in split_date_range(start_date, end_date, max_days=31):
        raw_xml = call_get_dam_price_period(
            start_date=chunk_start,
            end_date=chunk_end,
            in_eur=in_eur,
        )

        part_df = parse_to_simple_df(raw_xml)

        if not part_df.empty:
            all_parts.append(part_df)

    if not all_parts:
        return pd.DataFrame(columns=["DateTime", "Value"])

    df = pd.concat(all_parts, ignore_index=True)
    df = df.drop_duplicates(subset=["DateTime"]).sort_values("DateTime").reset_index(drop=True)
    return df
```

```python
chunks = split_date_range(START_DATE, END_DATE, max_days=31)
print("Chunks:")
for c in chunks:
    print(c)

df = load_dam_price_period_range(
    start_date=START_DATE,
    end_date=END_DATE,
    in_eur=IN_EUR,
)

print("Rows parsed:", len(df))
df.head(20)
```

```python
import pandas as pd
from datetime import date, datetime
from decimal import Decimal
import xml.etree.ElementTree as ET
from urllib.request import urlopen

ECB_HIST_URL = "https://www.ecb.europa.eu/stats/eurofxref/eurofxref-hist.xml"
NS = {"def": "http://www.ecb.int/vocabulary/2002-08-01/eurofxref"}


def _to_date(value):
    if isinstance(value, pd.Timestamp):
        return value.date()
    if isinstance(value, datetime):
        return value.date()
    if isinstance(value, date):
        return value
    return pd.to_datetime(value).date()


def _download_xml(url):
    with urlopen(url) as response:
        return ET.fromstring(response.read())


def load_czk_rates():
    root = _download_xml(ECB_HIST_URL)
    rates_by_date = {}

    for cube in root.findall(".//def:Cube", NS):
        time_value = cube.attrib.get("time")
        if not time_value:
            continue

        day = datetime.strptime(time_value, "%Y-%m-%d").date()

        for child in cube.findall("def:Cube", NS):
            if child.attrib.get("currency") == "CZK":
                rates_by_date[day] = Decimal(child.attrib["rate"])
                break

    return rates_by_date


def eur_to_czk_historic(amount_eur, rate_date, rates_by_date, previous_available=True):
    amount = Decimal(str(amount_eur))
    target_date = _to_date(rate_date)

    if target_date in rates_by_date:
        return amount * rates_by_date[target_date]

    if not previous_available:
        raise ValueError(f"No ECB rate for {target_date.isoformat()}.")

    available_dates = [d for d in rates_by_date if d <= target_date]
    if not available_dates:
        raise ValueError(f"No historic rate available on or before {target_date.isoformat()}.")

    nearest_date = max(available_dates)
    return amount * rates_by_date[nearest_date]


def convert_value_column_eur_mwh_to_czk_kwh(
    df,
    rates_by_date,
    datetime_col="DateTime",
    value_col="Value",
    previous_available=True,
):
    df = df.copy()

    df[value_col] = df.apply(
        lambda row: float(
            eur_to_czk_historic(
                amount_eur=row[value_col],
                rate_date=row[datetime_col],
                rates_by_date=rates_by_date,
                previous_available=previous_available,
            )
        ) / 1000,
        axis=1,
    )

    return df
```

```python
rates_by_date = load_czk_rates()
df = convert_value_column_eur_mwh_to_czk_kwh(df, rates_by_date)
print(df)
```

```python
deleted = delete_period_from_db(
    start_datetime=df["DateTime"].min(),
    end_datetime=df["DateTime"].max(),
    db_path=SQLITE_PATH,
    table_name=TABLE_NAME,
)


inserted = upsert_dataframe(df, SQLITE_PATH, TABLE_NAME)
print("Deleted:", deleted)
print("Inserted:", inserted)
```
